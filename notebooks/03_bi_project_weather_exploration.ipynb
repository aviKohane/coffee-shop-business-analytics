## 4. Missing Values Analysis

```python
cols_of_interest = [
    "temp", "temp_min", "temp_max", "feels_like",
    "humidity", "rain_1h", "snow_1h", "clouds_all",
    "weather_main","weather_description"
]

df_weather[cols_of_interest].isnull().sum()
```

### Key Observations

- No missing values detected in core weather variables such as temperature, humidity, and cloud coverage  
- Missing values are present in `rain_1h` and `snow_1h`  

These missing values are expected, as precipitation is only recorded when it occurs.

Overall, missing values do not indicate data quality issues.

## 6. Time Consistency Check

```python
df_weather["dt_iso"].nunique(), df_weather.shape[0]
```

## 7. Weather Variables Validation

```python
df_weather[["temp", "temp_min", "temp_max"]].describe()
```

### Key Observations

Overall, temperature data is consistent and reliable.

## 8. Temperature Consistency Check

```python
(df_weather["temp_max"] < df_weather["temp_min"]).sum()
```

No inconsistencies detected between minimum and maximum temperature values.

All records satisfy the condition:
temp_max ≥ temp_min

This confirms internal consistency of temperature data.

Overall, the dataset is reliable and ready for use in weather-based analysis.
