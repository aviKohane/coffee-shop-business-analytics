## 8. Hourly Granularity Validation

```python
df_calendar['date'] = pd.to_datetime(df_calendar['date'])
df_calendar['hour_check'] = df_calendar['time'].apply(lambda t: t.hour)

# 1) Check: exactly 24 rows per day
rows_per_day = df_calendar.groupby(df_calendar['date'].dt.date).size()

print("Rows per day -> min:", rows_per_day.min(), "| max:", rows_per_day.max())
bad_rows_days = rows_per_day[rows_per_day != 24].sort_values()
bad_rows_days.head(30)
```

### Hourly Granularity Validation

Each day contains exactly 24 records (min = 24, max = 24).

No anomalies were detected in the number of records per day.

This confirms that the dataset has a consistent hourly granularity, which is essential for time-based analysis in the BI model.
